 Cuando se utiliza para AND, la neurona debe activarse solo cuando todas las entradas sean 1, siendo el umbral (0):2, en la función de activación debe ser Y = 1 si X1.1 +X2+1 >2, si no Y=0.

Cuando se utiliza para OR, la neurona debe activarse cuando al menos una de las entradas sea 1, aquí las entradas van con los pesos , siendo el umbral (0): 1, en la función de activación debe ser Y = 1 si X1.1 +X2+1 >1, si no Y=0.

En cambio para NOT, la neurona tiene una entrada siempre de 1, aquí la neurona si tiene un sesgon en el umbral (0)=(0), Si X1=0 no hay detención, de lo contrario si X1=1 la neurona se apaga por completo, como lo comentamos en la clase con el apagador, solo tiene 2 funciones.

##Neurona de McCullock_Pitts

```python
import numpy as np
# Entradas para la neurona
X = np.array([
    [0, 0],
    [0, 1],
    [1, 0],
    [1, 1]
])
# Salidas
Y = np.array([0, 0, 0, 1])
# Pesos para las entradas
W = [1, 1]
# Umbral para la función de activación
umbral = 2
```

#Definimos la clase McCullochPitts

```python
class McCullochPitts:
    def __init__(self, weights, umbral):
        """
            Constructor de la neurona MCP.
            Guarda las variables
            weights -> vector de pesos iniciales
            umbral -> umbral para usar en la funcion de activación
        """
        self.weights = weights
        self.umbral = umbral

    def predict(self, X):
        """
            Calcula la salida de la neurona para cada fila de entradas X.
        """
        # Lista donde se guardarán las salidas
        predictions = []
        # Recorremos cada fila de entradas
        for i in range(X.shape[0]):
            # Calculamos la suma ponderada: w1*x1 + w2*x2
            weighted_sum = np.dot(self.weights, X[i])
            # Mostramos en consola la entrada y la suma ponderada
            print(f"Input: {X[i]}, Weighted Sum: {weighted_sum}")
            # Comparación con el umbral determinar si la neurona se activa o no
            if weighted_sum >= self.umbral:
                predictions.append(1)
            else:
                predictions.append(0)
        return predictions
```

#Usamos la clase McCullochPitts  para verificar los resultados.

```python
# Instanciamos la clase McCullochPitts
mcp = McCullochPitts(W, umbral)
# Calculamos las predicciones
predictions = mcp.predict(X)
# Mostramos el resultado final
print("Predictions:", predictions)
```
